1: Employee Performance Bonus Eligibility


```python
employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

# Find the highest performance score
highest_score = max(employees.values())

# Find all employees with the highest score
top_performers = [name for name, score in employees.items() if score == highest_score]

print("Top Performers Eligible for Bonus:", ", ".join(top_performers), f"(Score: {highest_score})")
```

2: Search Query Keyword Analysis


```python
query = "Buy mobile phone buy phone online"

# Convert to lowercase
query = query.lower()

# Split words
words = query.split()

# Count frequency
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

# Display keywords searched more than once
result = {word: count for word, count in frequency.items() if count > 1}

print(result)
```

3: Sensor Data Validation


```python
sensor_readings = [3, 4, 7, 8, 10, 12, 5]

valid_readings = []

for index, value in enumerate(sensor_readings):
    if value % 2 == 0:
        valid_readings.append((index, value))

print("Valid Sensor Readings (Hour, Value):")
print(valid_readings)
```

4: Email Domain Usage Analysis


```python
emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

domain_count = {}

# Count domains
for email in emails:
    domain = email.split("@")[1]
    domain_count[domain] = domain_count.get(domain, 0) + 1

total_emails = len(emails)

# Calculate and display percentage usage
for domain, count in domain_count.items():
    percentage = (count / total_emails) * 100
    print(f"{domain}: {int(percentage)}%")
```

 5: Sales Spike Detection


```python
sales = [1200, 1500, 900, 2200, 1400, 3000]
# Calculate average sales
average_sales = sum(sales) / len(sales)

# Detect spikes (more than 30% above average)
threshold = average_sales * 1.3

for day, value in enumerate(sales, start=1):
    if value > threshold:
        print(f"Day {day}: {value}")
```
