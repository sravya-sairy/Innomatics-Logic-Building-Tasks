
Problem Statement 1:Smart Parking Lot Management System


```python
def parking_lot(capacity, logs):
    parked = 0
    peak = 0

    for log in logs:
        if log == "IN":
            parked += 1
        elif log == "OUT":
            parked -= 1

        if parked > peak:
            peak = parked

    print("Currently Parked Vehicles:", parked)

    if parked > capacity:
        print("Parking Status: Full (Alert!)")
    else:
        print("Parking Status: Available")

# Sample Input
parking_capacity = 50
vehicle_logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

parking_lot(parking_capacity, vehicle_logs)
```

 Problem Statement 2:Online Food Delivery Time Estimator


```python
def delivery_time(distance, traffic, weather):
    time = distance * 5   # base time: 5 minutes per km

    if traffic == "High":
        time += 15
    elif traffic == "Medium":
        time += 8

    if weather == "Rainy":
        time += 10

    print("Estimated Delivery Time:", time, "minutes")

# Sample Input
delivery_time(8, "High", "Rainy")
```

 Problem Statement 3:Movie Theatre Seat Occupancy Analyzer


```python
def seat_occupancy(total_seats, booked_seats):
    booked = len(booked_seats)
    occupancy = (booked / total_seats) * 100

    print("Occupancy:", int(occupancy), "%")

    if occupancy == 100:
        status = "Housefull"
    elif occupancy >= 75:
        status = "Almost Full"
    else:
        status = "Seats Available"

    print("Show Status:", status)

# Sample Input
total = 200
booked = [1] * 150

seat_occupancy(total, booked)
```

 Problem Statement 4:Cloud Server Load Classification System


```python
def server_load(cpu_readings):
    total = 0

    for cpu in cpu_readings:
        total += cpu

    average = total / len(cpu_readings)

    print("Average CPU Load:", int(average), "%")

    if average < 50:
        status = "Normal"
    elif average <= 80:
        status = "Warning"
    else:
        status = "Critical"

    print("Server Status:", status)

# Sample Input
cpu_data = [45, 60, 70, 85, 90]

server_load(cpu_data)
```

 Problem Statement 5:Smart Classroom Resource Usage Monitor


```python
def resource_monitor(resources):
    overused = []

    for resource, hours in resources.items():
        if hours > 6:
            overused.append(resource)

    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")

# Sample Input
usage = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

resource_monitor(usage)
```

 Problem Statement 6: Online Event Registration Capacity Controller


```python
def event_registration(capacity, registrations):
    if registrations <= capacity:
        print("Confirmed Registrations:", registrations)
        print("Waitlisted Users: 0")
        print("Registration Status: Open")
    else:
        waitlist = registrations - capacity
        print("Confirmed Registrations:", capacity)
        print("Waitlisted Users:", waitlist)
        print("Registration Status: Closed")

# Sample Input
event_registration(100, 105)
```
