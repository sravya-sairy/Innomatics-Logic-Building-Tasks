Problem 1: Mobile Recharge Validation System

```python
def validate_recharge():
    valid_plans = [199, 299, 399, 599]

    while True:
        amount = int(input("Enter recharge amount: ₹"))

        if amount < 50:
            print(" Recharge amount must be at least ₹50")
        elif amount not in valid_plans:
            print(" Invalid plan. Available plans:", valid_plans)
        else:
            print(" Recharge successful for ₹", amount)
            break

# Function call
validate_recharge()
```

Problem 2: Inventory Reorder Alert System

```python
def check_inventory(products):
    for item, stock in products.items():
        if stock < 15:
            print(item, "→ Reorder Alert")
        else:
            print(item, "→ Stock OK")

# Input dictionary
inventory = {
    "Rice": 20,
    "Sugar": 10,
    "Oil": 8,
    "Salt": 25
}

check_inventory(inventory)
```

 Problem 3: Student Result Processing System


```python
def student_result(marks):
    total = 0

    for mark in marks:
        total += mark

    average = total / len(marks)

    if average >= 50:
        return "Pass"
    else:
        return "Fail"

# Example input
marks_list = [60, 55, 40, 70]
result = student_result(marks_list)

print("Result:", result)
```

Problem 4: Cab Fare Estimator with Retry Option

```python
def calculate_fare(distance, peak_hour):
    base_fare = 50
    fare = base_fare + (distance * 12)

    if peak_hour == "yes":
        fare += fare * 0.25

    return fare

while True:
    km = float(input("Enter distance (km): "))
    peak = input("Peak hour? (yes/no): ").lower()

    total_fare = calculate_fare(km, peak)
    print("Total Fare: ₹", total_fare)

    retry = input("Do you want to calculate again? (yes/no): ").lower()
    if retry != "yes":
        break
```

Problem 5: Employee Attendance Eligibility Checker


```python
def check_attendance(attendance):
    present_days = 0

    for day in attendance:
        if day == "P":
            present_days += 1

    percentage = (present_days / len(attendance)) * 100

    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"

# Example attendance list
attendance_list = ["P", "P", "A", "P", "P", "A", "P"]

status = check_attendance(attendance_list)
print("Attendance Status:", status)
```

Problem 6: Password Strength Checker


```python
def check_password(password):
    special_chars = "@#$"
    has_digit = False
    has_special = False

    if len(password) < 8:
        return "Weak Password"

    for ch in password:
        if ch.isdigit():
            has_digit = True
        if ch in special_chars:
            has_special = True

    if has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak Password"

# Input password
pwd = input("Enter password: ")
print(check_password(pwd))
```
