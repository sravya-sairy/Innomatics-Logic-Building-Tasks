Problem 1: Social Media – Post Engagement Analyzer


```python
def analyze_engagement(likes):
    total_likes = 0
    for like in likes:
        total_likes += like

    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"

    print("Total Likes:", total_likes)
    print("Post Status:", status)

# Example input
likes_list = [200, 350, 300, 200]
analyze_engagement(likes_list)
```

Problem 3: Agriculture – Rainfall Adequacy Checker


```python
def rainfall_check(rainfall, required):
    total = 0
    for r in rainfall:
        total += r

    average = total / len(rainfall)

    if average >= required:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"

    print("Average Rainfall:", average)
    print("Rainfall Status:", status)

# Example input
rainfall_data = [70, 75, 80, 65, 70]
rainfall_check(rainfall_data, 70)
```

Problem 4: Social Media – Duplicate Account Detection


```python
def check_duplicates(usernames):
    if len(usernames) != len(set(usernames)):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")

# Example input
users = ["ram", "sita", "ram", "john"]
check_duplicates(users)
```

Problem 6: Agriculture – Premium Crop Price Filter


```python
def premium_crops(prices):
    premium = []

    for price in prices:
        if price > 2000:
            premium.append(price)

    print("Premium Crops:", premium)

# Example input
crop_prices = [1500, 2500, 1800, 3200, 2100]
premium_crops(crop_prices)
```

Problem 7: System Monitoring – Application Health Checker


```python
def system_health(errors):
    if errors == 0:
        status = "Healthy"
    elif errors <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"

    print("Error Count:", errors)
    print("System Status:", status)

# Example input
system_health(7)
```

Problem 9: E-Learning – Student Attendance Eligibility System


```python
def attendance_eligibility(attendance):
    present = 0

    for day in attendance:
        if day == "P":
            present += 1

    percentage = (present / len(attendance)) * 100

    if percentage >= 75:
        status = "Eligible"
    else:
        status = "Not Eligible"

    print("Attendance Percentage:", percentage)
    print("Exam Eligibility:", status)

# Example input
attendance_list = ["P", "P", "A", "P", "P"]
attendance_eligibility(attendance_list)
```

Problem 10: Smart Electricity Bill Analyzer


```python
def electricity_bill(units):
    bill = 0

    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"

    return bill, usage

# Example input
total_bill, usage_status = electricity_bill(250)
print("Total Bill:", total_bill)
print("Usage Level:", usage_status)
```
